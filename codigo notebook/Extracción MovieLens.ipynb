# Extractor de data MovieLens

```python
import requests
import zipfile
from pathlib import Path
import pandas as pd
```

### Extraemos los datos MovieLens y guardamos la data en bruto para precesamiento


```python
url = "https://files.grouplens.org/datasets/movielens/ml-32m.zip"

ruta_zip = Path(r"D:\cinematch\data\bronze\ml-32m.zip")
destino = Path(r"D:\cinematch\data\bronze")

ruta_zip.parent.mkdir(
    parents=True,
    exist_ok=True
)

with requests.get(url, stream=True) as respuesta:
    respuesta.raise_for_status()

    with open(ruta_zip, "wb") as archivo:
        for bloque in respuesta.iter_content(chunk_size=1024 * 1024):
            if bloque:
                archivo.write(bloque)

print("Descarga completada.")

with zipfile.ZipFile(ruta_zip, "r") as archivo_zip:
    archivo_zip.extractall(destino)

print("MovieLens 32M extraído correctamente.")
```

### Análisis de datos

```python
for archivo in archivos:
    if archivo.is_file():
        tamano_mb = archivo.stat().st_size / (1024 ** 2)

        print(
            archivo.name,
            "→",
            round(tamano_mb, 2),
            "MB"
        )
```
