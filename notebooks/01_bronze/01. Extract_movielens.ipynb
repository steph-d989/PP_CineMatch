# EXTRAER DATA DE MOVIELENS

```python
import requests
import zipfile
import sys
from pathlib import Path
import pandas as pd
```

```python
PROJECT_ROOT = Path.cwd()

while not (PROJECT_ROOT / "src").exists() and PROJECT_ROOT.parent != PROJECT_ROOT:
    PROJECT_ROOT = PROJECT_ROOT.parent

sys.path.append(str(PROJECT_ROOT))

from src.utils.paths import BRONZE_MOVIELENS_DIR
```

### 1. Identificar la ruta del proyecto y extraer la data


```python
url = "https://files.grouplens.org/datasets/movielens/ml-32m.zip"

ruta_zip = BRONZE_MOVIELENS_DIR / "ml-32m.zip"
destino = BRONZE_MOVIELENS_DIR

with requests.get(url, stream=True) as respuesta:
    respuesta.raise_for_status()

    with open(ruta_zip, "wb") as archivo:
        for bloque in respuesta.iter_content(chunk_size=1024 * 1024):
            if bloque:
                archivo.write(bloque)

print("Descarga completada.")

with zipfile.ZipFile(ruta_zip, "r") as archivo_zip:
    archivo_zip.extractall(destino)

print("MovieLens 32M extraído correctamente.")
```

### 2. Verificación de archivos extraídos

Los datos que están en la carpeta son los siguientes:

- links.csv : Relacion con IMDb
- movies.csv: Información de la película
- ratings.csv : Calificación de usuarios
- tags.csv : Etiquetas dadas por usuarios

```python
ruta_movielens = BRONZE_MOVIELENS_DIR / "ml-32m"

archivos = list(ruta_movielens.iterdir())

for i in archivos:
    print(i.name)
```

### 3. Revisar tamaño de archivos

```python
for archivo in archivos:
    if archivo.is_file():
        tamano_mb = archivo.stat().st_size / (1024 ** 2)

        print(
            archivo.name,
            "->",
            round(tamano_mb, 2),
            "MB"
        )
```

### 4. Validacion de ingesta

```python
ruta_movies = ruta_movielens / "movies.csv"
ruta_ratings = ruta_movielens / "ratings.csv"
ruta_tags = ruta_movielens / "tags.csv"
ruta_links = ruta_movielens / "links.csv"

movies = pd.read_csv(ruta_movies)
ratings = pd.read_csv(ruta_ratings)
tags = pd.read_csv(ruta_tags)
links = pd.read_csv(ruta_links)
```

### 5. Mostrar cantidad de datos

```python
print("movies:", movies.shape)
print("ratings:", ratings.shape)
print("tags:", tags.shape)
print("links:", links.shape)
```
